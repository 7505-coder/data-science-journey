### Loading the Libraries and Dataset

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')


df = pd.read_csv('IPL.csv')
df.head()
```

**Now, Here comes some Basic Questions**

1. Which team won the most matches?

```python
match_wins = df['match_winner'].value_counts()
match_wins
```

```python
Most_Matches_Won = sns.barplot(x = match_wins.values, y = match_wins.index,palette='rainbow')
plt.title('Most_Matches_Won')
```

2. Toss Decision Trends

```python
Toss_Decision = df['toss_decision'].value_counts()
Toss_Decision
```

```python
sns.countplot(x = df['toss_decision'])
plt.title('Toss Decision Trends')
```

3. Toss Winner vs Match Winner

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage = (count * 100)/df.shape[0]
percentage
print(f"There is only {percentage} chance of winning a match if you win the toss")
```

4. How do teams win? (Runs vs Wickets)

###  Key Player Performances

1 Most "Player of the Match" Awards

```python
count = df['player_of_the_match'].value_counts().head(10)
count
```

```python
sns.barplot(y = count.index, x = count.values,palette="mako")
```

2 Top Scorers

```python
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
high
```

```python
high.plot(kind = 'barh')
```

10 Best Bowling Figures

```python
df.head()
```

```python
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x :x.split('--')[0])
df['highest_wickets'] = df['highest_wickets'].astype(int)
top_bollers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
top_bollers
```

```python
sns.barplot(y = top_bollers.index, x= top_bollers.values)
```
