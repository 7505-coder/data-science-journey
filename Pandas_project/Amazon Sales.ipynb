```python
import numpy as np
import pandas as pd

df = pd.read_csv('amazon.csv')
df.head()
```

1. Check the data types of all columns. Which ones should be numeric but are currently stored as strings?

```python
df.dtypes
#discounted_price,actual_price,discount_percentage and rating shold be in float and rating_count should be in integer
```

2. cleaning discounted_price and actual_price (stripping the ₹ symbol and commas, then converting to float)

```python
df['discounted_price']= df['discounted_price'].str.replace('₹','').str.replace(',','')
df['actual_price']= df['actual_price'].str.replace(',','').str.replace('₹','')
df[['discounted_price','actual_price']] = df[['discounted_price','actual_price']].astype(float)
```

```python
df[['discounted_price','actual_price']].dtypes
```

3. Clean discount_percentage: remove the % sign and convert to float.

```python
df['discount_percentage']= df['discount_percentage'].str.replace('%','')
df['discount_percentage'] = df['discount_percentage'].astype(float)
```

```python
df['discount_percentage'].dtypes
```

4. Clean rating_count: remove commas and convert to numeric. What happens to the 2 missing values — check what's actually in those rows before deciding whether to drop or fill them.

```python
df[df['rating_count'].isnull()]
df = df.dropna(subset=['rating_count'])
```

```python
df['rating_count']= df['rating_count'].str.replace(',','')
df['rating_count'] = df['rating_count'].astype(int)
```

```python
df['rating_count'].dtypes
```

5. Try converting rating to numeric using pd.to_numeric(). You'll likely hit an error — find out which row(s) are causing it and decide how to handle it.

```python
df['rating'] = pd.to_numeric(df['rating'])
```

```python
df['rating'] = pd.to_numeric(df['rating'],errors= 'coerce')
df[df['rating'].isnull()]
df = df.dropna(subset='rating')
```

16. Find the average length (word count) of review_content — does a longer review correlate with a higher or lower rating?

```python
df.head()
df['review_word_count']= df['review_content'].str.split().str.len()
df['review_word_count'].mean().round(2)
df['review_word_count'].describe()
```

```python
corelation = df['review_word_count'].corr(df['rating'])
corelation
```

17. Verify your cleaned discount_percentage roughly matches 1 - (discounted_price / actual_price) — do the numbers align, or is there a mismatch worth investigating?

```python
df['implied_discount'] = (1 - df['discounted_price'] / df['actual_price']) * 100
```

```python
df['discount_diff'] = df['discount_percentage'] - df['implied_discount']
```
