20. Find the average movie duration (in minutes) per rating category.

```python
movies = df[df['type'] == 'Movie'].copy()
movies = movies.dropna(subset=['duration'])
movies['duration_mins'] = movies['duration'].str.replace(' min', '').astype(int)
```

```python
avg_movie_duration = movies.groupby('rating')['duration_mins'].mean()
avg_movie_duration.sort_values(ascending=False).round()
```

Section 6: Advanced / Pivot

23. Create a pivot table showing count of type (Movie/TV Show) by release_year for the last 10 years.

```python
recent_years = df['release_year'].sort_values(ascending=False).unique()[:10]
print(recent_years)
```

```python
recent_df = df[df['release_year'].isin(recent_years)]
print(recent_df.shape)
```

```python
pivot = recent_df.pivot_table(index='release_year', columns='type', values='title', aggfunc='count')
print(pivot)
```
