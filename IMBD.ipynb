Section 2: Cleaning (Q6–10)

6. Find how many rows have a missing Revenue value.

```python
df['Gross'].isnull().sum()
```

7. Fill missing Revenue values with the median revenue.

```python
df['Gross'] = df['Gross'].str.replace(',', '').astype(float)
```

```python
df['Gross'] = df['Gross'].fillna(df['Gross'].median())
df['Gross']
```

8. Drop rows where Rating is missing (if any).

```python
df['IMDB_Rating'] = df['IMDB_Rating'].dropna()
```

9. Check for and remove duplicate rows.

```python
df = df.drop_duplicates()
```

10. Convert the Runtime column to integer type if it's stored as text (e.g., "142 min" → 142).

```python
df['Runtime'] = df['Runtime'].str.replace('min','').astype(int)
```

```python
df['Runtime'].dtype
```

Section 5: GroupBy & Aggregation (Q19–23)

19. Group by Year and find the average rating per year.

```python
df.groupby('Released_Year')['IMDB_Rating'].mean()
```

20. Group by Director and find who has the most movies in the dataset.

```python
df.groupby('Director')['Series_Title'].max()
```

21. Group by Genre and find the average revenue per genre.

```python
df_exploded = df.assign(Genre=df['Genre'].str.split(', ')).explode('Genre')
df_exploded.groupby('Genre')['Gross'].mean().round()
```

Section 6: Advanced/Combined (Q24–25)

24. Find the director with the highest average revenue (only consider directors with 3+ movies).

```python
dirt_stats = df.groupby('Director')['Gross'].agg((['mean','count']))
dirt_stats = dirt_stats[dirt_stats['count']>=3]
dirt_stats['mean'].idxmax()
```

25. Create a new column Profitability (if Budget column exists) or Decade (from Year, e.g., 2010s, 2000s) and group movies by it to compare average ratings.

```python
df[df['Released_Year'].str.isnumeric() == False]
```

```python
df = df[df['Released_Year'].str.isnumeric() == True]
df['Released_Year'] = df['Released_Year'].astype(int)
```

```python
df['Decade'] = (df['Released_Year'] // 10) * 10
df.groupby('Decade')['IMDB_Rating'].mean().sort_values(ascending=False)
```
